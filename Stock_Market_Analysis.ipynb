# Checking files which contains the null values

```python
import os
import pandas as pd

stocks_folder = os.path.join(path, 'stocks')

# Dictionary to store filenames and their respective null counts
files_with_nulls = {}

print("Scanning files for null values. This may take a moment...")

# Loop through every file in the directory
for filename in os.listdir(stocks_folder):
    if filename.endswith(".csv"):
        file_path = os.path.join(stocks_folder, filename)

        # Load the CSV file
        df = pd.read_csv(file_path)

        # Calculate the total number of missing values in the dataset
        total_nulls = df.isnull().sum().sum()

        # Record the file if it contains any null values
        if total_nulls > 0:
            files_with_nulls[filename] = total_nulls
            print(f"{filename}: {total_nulls} null values")

print(f"\nScan complete. Found {len(files_with_nulls)} files with null values.")
```
